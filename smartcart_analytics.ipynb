# ⭐ Imputing Missing Values in the Rating Column

The **rating** column is one of the most important features in this dataset because it represents customer satisfaction and product quality. During data preprocessing, some records contained missing or improperly formatted rating values.

Before imputing missing values, the rating values were cleaned by removing the text **"out of 5 stars"** and converting the column to a numeric (`float`) data type.

After the column was converted to numeric format, missing values were imputed using the **median rating**. The median was selected because it is less affected by outliers and provides a more robust estimate for skewed data.

### 🎯 Objective

- ⭐ Remove unwanted text from the rating column.
- 🔄 Convert ratings to numeric (`float`) format.
- 📊 Impute missing values using the median.
- ✅ Ensure the column is complete and ready for analysis and visualization.

> **Result:** The cleaned **rating** column now contains only numeric values (e.g., **4.6**, **4.3**, **4.8**) with no missing values, making it suitable for statistical analysis and interactive visualizations.

````python
# -----------------------------
# Clean Rating Column
# -----------------------------

# Remove " out of 5 stars"
df["rating"] = df["rating"].str.replace(" out of 5 stars", "", regex=False)

# Convert to float
df["rating"] = pd.to_numeric(df["rating"], errors="coerce")

# Fill missing values with mean
df["rating"] = df["rating"].fillna(df["rating"].mean())

# Verify
print(df["rating"].head(10))
print("\nData Type:", df["rating"].dtype)
print("\nMissing Values:", df["rating"].isnull().sum())
````

# 📝 Cleaning the Number of Reviews Column

The **number_of_reviews** column was cleaned by removing unnecessary characters and converting the values into a numeric format. Missing values were handled appropriately to ensure accurate analysis and visualization.

**Result:** The column now contains only numeric review counts and is ready for Exploratory Data Analysis (EDA).

# 👁️ Display number_of_reviews values before cleaning

````python
df["number_of_reviews"].head(10)
````

# Imputing data in number_of_reviews column

````python
# -----------------------------
# Clean Number of Reviews Column
# -----------------------------

# Remove commas
df["number_of_reviews"] = df["number_of_reviews"].astype(str).str.replace(",", "", regex=False)

# Convert to numeric
df["number_of_reviews"] = pd.to_numeric(df["number_of_reviews"], errors="coerce")

# Fill missing values with median
df["number_of_reviews"] = df["number_of_reviews"].fillna(df["number_of_reviews"].median())

# Convert to integer
df["number_of_reviews"] = df["number_of_reviews"].astype(int)

# Verify
print(df["number_of_reviews"].head(10))
print("\nData Type :", df["number_of_reviews"].dtype)
print("Missing Values :", df["number_of_reviews"].isnull().sum())
````

# 📦 Imputing Data in the Bought in Last Month Column

The **bought_in_last_month** column was cleaned by extracting only the numeric purchase count, removing unnecessary text (e.g., "bought in past month"), and converting the values into a numeric format. Missing values were then imputed with **0** to represent products with no recorded monthly purchases.

**Result:** The column now contains only numeric values and is ready for analysis and visualization.

# 📦 Display bought_in_last_month values before cleaning

````python
# -----------------------------
# Bought in Last Month (Before Cleaning)
# -----------------------------

print("Before Cleaning:")
display(df["bought_in_last_month"].head(10))

print("\nMissing Values:", df["bought_in_last_month"].isnull().sum())
print("Data Type:", df["bought_in_last_month"].dtype)
````

# ✨Cleaning & Imputing


````python
# ------------------------------
# Clean Bought in Last Month Column
# ------------------------------

# Extract only the numeric part (e.g., 100+, 50+, 2K+)
df["bought_in_last_month"] = (
    df["bought_in_last_month"]
    .astype(str)
    .str.extract(r'(\d+K?|\d+)\+?')[0]
)

# Convert K values to numbers
df["bought_in_last_month"] = (
    df["bought_in_last_month"]
    .str.replace("K", "000", regex=False)
)

# Convert to numeric
df["bought_in_last_month"] = pd.to_numeric(
    df["bought_in_last_month"],
    errors="coerce"
)

# Fill missing values with 0
df["bought_in_last_month"] = (
    df["bought_in_last_month"]
    .fillna(0)
    .astype(int)
)

# Display after cleaning
print("\nAfter Cleaning:")
display(df["bought_in_last_month"].head(10))

print("\nMissing Values:", df["bought_in_last_month"].isna().sum())
print("Data Type:", df["bought_in_last_month"].dtype)
````

# 🚨 Outlier Detection

Outlier detection helps identify unusually high or low values in numerical columns that may affect data analysis and model performance.

In this project, the **Interquartile Range (IQR)** method is used to detect outliers before performing Exploratory Data Analysis (EDA).

### 🎯 Objectives

- Detect extreme values in numerical features.
- Evaluate data quality before visualization.
- Determine whether outliers represent genuine business cases or data errors.

> **Note:** Outliers are detected but not removed, as they may represent real-world Amazon products such as premium-priced items or highly popular products with exceptionally high review counts.

````python
for col in df.select_dtypes(include="number").columns:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = ((df[col] < lower) | (df[col] > upper)).sum()

    print(f"{col}: {outliers} outliers")
````

# 📌 Outlier Handling

## Overview

Outliers are extreme values that differ significantly from the rest of the dataset. In this project, outliers were identified during data exploration to understand the distribution of numerical features.

### Approach Used

- Outliers were detected using statistical methods.
- Since this dataset contains real Amazon product information, high prices and a large number of reviews represent genuine product characteristics rather than data errors.
- Therefore, no outliers were removed or modified to preserve the original business information and maintain data integrity.

### Conclusion

The identified outliers were retained because they reflect real-world variations in Amazon product sales and customer behavior.

````python
# Remove invalid ratings (ratings should be between 1 and 5)
df = df[(df["rating"] >= 1) & (df["rating"] <= 5)]

# Remove invalid prices (price cannot be negative)
df = df[df["current/discounted_price"] >= 0]

# Remove invalid review counts (reviews cannot be negative)
df = df[df["number_of_reviews"] >= 0]

print("Outliers handled successfully.")
````

## 📌 Introduction

Correlation analysis is used to measure the strength and direction of the relationship between numerical variables in the dataset. It helps identify whether two variables increase together, decrease together, or have little to no relationship.

## 🎯 Objective

- Analyze relationships between numerical features.

- Support data-driven business decisions.
- Detect patterns useful for further analysis.

## 📊 Visualization

**Chart Type:** Interactive Correlation Heatmap  
**Library:** Plotly  
**Purpose:** Display the correlation coefficients between numerical variables using an interactive color-coded heatmap.

````python
import plotly.express as px
# Select only numerical columns
numeric_df = df.select_dtypes(include="number")

# Calculate correlation matrix
correlation = numeric_df.corr()

# Create Interactive Heatmap
fig = px.imshow(
    correlation,
    text_auto=".2f",
    color_continuous_scale="RdBu_r",
    aspect="auto"
)

# Update Layout
fig.update_layout(
    title={
        "text": "<b>Correlation Heatmap of Numerical Features</b>",
        "x": 0.5,
        "xanchor": "center"
    },
    xaxis_title="Numerical Features",
    yaxis_title="Numerical Features",
    width=900,
    height=700,
    template="plotly_white"
)

# Display Heatmap
fig.show()
````

## 💡 Graph Overview

- The heatmap displays the correlation coefficient between every pair of numerical variables.
- Values close to **+1** indicate a strong positive relationship.
- Values close to **-1** indicate a strong negative relationship.
- Values near **0** indicate little or no linear relationship.
- Darker colors represent stronger correlations, while lighter colors indicate weaker relationships.

Correlation analysis helps identify relationships among important business metrics. # 💬 Customer Reviews Distribution

## 📌 Introduction

Customer reviews indicate product popularity and customer engagement. Since review counts are highly skewed, a logarithmic transformation is applied to improve visualization and better understand the overall distribution.

## 🎯 Objective

- Analyze the distribution of customer reviews.
- Reduce the effect of extremely large review counts.
- Visualize customer engagement using an interactive Plotly chart.

## 📊 Visualization

**Chart Type:** Smooth Distribution Curve (Log Scale)

**Library:** Plotly

**Purpose:** Display the distribution of customer reviews using a smooth interactive line chart without requiring SciPy.

````python
import numpy as np
import plotly.graph_objects as go

# Remove missing values
reviews = df["number_of_reviews"].dropna()

# Log Transformation
log_reviews = np.log1p(reviews)

# Create histogram bins
counts, bins = np.histogram(log_reviews, bins=40)

# Bin centers
centers = (bins[:-1] + bins[1:]) / 2

# Create Smooth Line
fig = go.Figure()

fig.add_trace(
    go.Scatter(
        x=centers,
        y=counts,
        mode="lines",
        line=dict(color="royalblue", width=4, shape="spline"),
        fill="tozeroy",
        fillcolor="rgba(65,105,225,0.25)",
        name="Customer Reviews"
    )
)

fig.update_layout(
    title={
        "text": "<b>Distribution of Customer Reviews (Log Scale)</b>",
        "x": 0.5
    },
    xaxis_title="Log(Number of Reviews)",
    yaxis_title="Number of Products",
    template="plotly_white",
    width=950,
    height=550,
    hovermode="x unified"
)

fig.show()
````

## 💡 Graph Overview

### Key Observations

- The smooth distribution curve highlights how customer reviews are spread across products.
- Most products have relatively fewer reviews.
- A small number of products receive exceptionally high customer engagement.

### Business Insight

The log-transformed distribution makes it easier to compare products with different review counts. This visualization helps identify customer engagement patterns and supports better business analysis.

## 🕸️ Top 10 Most Purchased Products (Radar Chart)

This radar chart visualizes the top 10 most purchased products based on the `bought_in_last_month` column. Each axis represents a product, while the distance from the center indicates the number of purchases. This chart provides a unique and interactive way to compare product popularity.

````python
import pandas as pd
import plotly.graph_objects as go

# Clean purchase count
df["bought_in_last_month"] = (
    df["bought_in_last_month"]
    .astype(str)
    .str.replace(",", "", regex=False)
    .str.replace("+", "", regex=False)
    .str.replace("K", "000", regex=False)
)

df["bought_in_last_month"] = pd.to_numeric(
    df["bought_in_last_month"], errors="coerce"
)

# Top 10 Most Purchased Products
top10 = (
    df[["title", "bought_in_last_month"]]
    .dropna()
    .sort_values("bought_in_last_month", ascending=False)
    .head(10)
)

# Shorten product names
top10["Product"] = top10["title"].apply(
    lambda x: x[:25] + "..." if len(x) > 25 else x
)

# Radar Chart
fig = go.Figure()

fig.add_trace(go.Scatterpolar(
    r=top10["bought_in_last_month"],
    theta=top10["Product"],
    fill='toself',
    line=dict(color='royalblue', width=3),
    marker=dict(size=8, color='crimson'),
    hovertemplate="<b>%{theta}</b><br>Purchased: %{r}<extra></extra>"
))

fig.update_layout(
    title="Top 10 Most Purchased Products",
    title_x=0.5,
    template="plotly_white",
    width=800,
    height=800,
    polar=dict(
        radialaxis=dict(
            visible=True,
            showline=True,
            gridcolor="lightgray"
        )
    ),
    showlegend=False
)

fig.show()
````

### 📊 Graph Overview

- **Chart Type:** Radar (Spider) Chart
- **Columns Used:** `title`, `bought_in_last_month`
- **Library:** Plotly Graph Objects

**Key Insights:**
- Displays the top 10 most purchased products.
- Larger values extending from the center indicate higher purchase counts.
- Hovering over each point shows the product name and purchase count.
- Provides a visually engaging comparison of product popularity.
